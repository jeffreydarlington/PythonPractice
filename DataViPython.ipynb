```python
import pandas as pd

df = pd.read_csv("InterviewQs(Sheet1).csv")

df_sorted = df.sort_values(by=["Department", "Salary"], ascending=[True, False])

top_2_per_department = (
    df_sorted
    .groupby("Department")
    .head(2)
    [["Employee_ID", "Name", "Department", "Salary"]]
)

print(top_2_per_department)
```
